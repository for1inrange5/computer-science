**Panduan Penulisan kode pada Python** \
Keterbacaan kode gak ada dampak kepada kemampuan komputer untuk memahami program, karena pada dasarnya komputer itu literal, tapi keterbacaan kode itu dapat menguntungkan manusia.

Disini gak banyak panduan cuma yah gaya yang paling umum dipakai

```python
# Penulisan variabel yang jelas
umur = 18
nama = "Panji Wicaksono"
kuliah = "D4 TE FV UNY"
# Kalo variabel butuh spasi, di kasih _ dan kecil semua
# contoh : nama_lengkap
# itu namanya snake case

# Mencetak nama umur dan kuliahnya dimana
print(f"{nama} {umur} {kuliah}")
# Cara baru, Formating String Literals
# Masukin variabel langsung ke dalam fungsi
# jangan lupa pake f di depan
```

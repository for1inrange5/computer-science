#### If bersarang, atau kondisi bersarang

Ini if di dalam if, kayak jika benar di lanjutkan ke kondisi berikutnya, if bercabang bisa super banyak, ada contoh di bawah ini:

**Studi Kasus 1. Cek Provinsi dan Kabupaten**

```python
# Input provinsi
provinsi = input("Dimana provinsimu? ")

# Cek kondisi provinsi
if provinsi == "Yogyakarta":
    print("Kamu orang yogyakarta!")
    kabupaten = input("Dimana kabupatenmu? ")
    
    # jika provinsi yogyakarta, yang dimana itu true, lanjut ke kabupaten
    if kabupaten == "Gunungkidul":
        print("Kamu orang Gunungkidul!")
    elif kabupaten == "Bantul":
        print("Kamu orang mbantul!")
    elif kabupaten == "Sleman":
        print("Kamu orang Sleman!")
    elif kabupaten == "Kulonprogo":
        print("Kamu orang Kulonprogo")
    elif kabupaten == "Yogyakarta":
        print("Kamu orang kota Yogyakarta! ")
    else:
        print("Kamu bukan orang yogyakarta, kamu bohong!")
else:
    print("Sorry i'didnt know about that")

print("Terimakasih sudah mencoba programku!")
```

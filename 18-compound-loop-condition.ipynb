### **Compound loop conditions / Kondisi loop gabungan**

Bagaimana jika kita hanya ingin menginginkan `while loop` ketika kedua kondisi adalah `True`?

**Loop termination**
Kita bisa menggunakan setiap ekspresi yang menghasilkan `True` atau `False` yang menjadi kondisi loop. contoh:

```python
import random

p1_wins = 0
p2_wins = 0

# Pemain pertama yang memenangkan 3 putaran akan memenangkan permainan.
while p1_wins < 3 and p2_wins < 3:
    winner = random.randint(1, 2)

    if winner == 1:
        print("Player 1 wins!")
        p1_wins = p1_wins + 1
    else:
        print("Player 2 wins!")
        p2_wins = p2_wins + 1
```

**Kondisi loop dengan`or`**

Sebuah perulangan dengan operator `or` akan berulang selama setidaknya satu kondisi bernilai `True`. Perulangan tersebut akan berakhir hanya jika kedua kondisi bernilai `False`.

```python
import random

battery = random.randint(1, 5)
print("Starting at " + str(battery) + " battery.")

gas_tank = random.randint(1, 5)
print("Starting at " + str(gas_tank) + " gas.")

# Kendaraan hibrida dapat menggunakan tenaga baterai maupun bahan bakar.
while battery > 0 or gas_tank > 0:

    # Lebih baik menggunakan baterai daripada bahan bakar, jika daya baterainya masih tinggi.
    if battery > gas_tank:
        print("Battery!")
        battery = battery - 1
    else:
        print("Gas!")
        gas_tank = gas_tank - 1
```

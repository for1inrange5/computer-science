## Macam macam error di Python

Dalam python ada 3 error yang bisa kita pahami, di antaranya Syntax Error, Runtime Error, dan Logic error

### Mengapa harus belajar error? kenapa emang?

sebenernya kalo lu jago ga usah belajar beginian juga gapapa, ada AI juga, tapi ini membuat lu mengenali error lebih baik, bukan bikin lu langsung jago debuging, oke lanjut!

#### 1. Syntax Error

nah ini error yang paling umum, terkadang kita suka salah ketik biasanya kurang `)`, `}` atau ngetik yang ga ada di aturan python

```python
# Contoh error
print("Hello Wolrd)
```

#### 2. Name Error
Name error itu salah satu jenis exception di python kalo intrepreter tidak bisa menemukan nama yang sedang dirujuk `print` jadi `prinr`

```python
# Contoh Error
prinl("Hello World")
```

#### 3. Runtime Error
Runtime error terjadi karena program yang lu eksekusi mengalami kesalahan, udah bener sintaksnya, tapi ada kesalahan di programnya, contoh

```python
# Contoh Error
print(0 / 0);
# Gak ada sintaks error, tapi error di programnya
```

#### 4. Logic Error
Nah ini error yang sulit buat di identifikasi, kenapa? karena sintaks ga ada yang error, program juga jalan, tapi outputnya beda sama yang kamu harepin contoh

```python
# Contoh menghitung luas
# Kalo biasanya p * l, eh malah p + l
luas = 10 + 11
# padahal 10 * 11
print(luas)
```

#### 5. Error Exception, Bonus
Exception adalah sebuah kondisi atau kejadian yang muncul ketika program sedang dieksekusi dan mengganggu alur normal eksekusi program.

**While loop**

ini looping di dalam kondisi, like kalo n < 5, kalo udah memenuhi syaratnya, dia bakal berhenti

```python
n = 1
j = 5

while n <= j:
    print(n)
    n = n + 1
```

kenapa berhenti? karena n itu 5, kalo n lebih kecil sama dengan 6, itu salah, jadi dia berhenti di 5

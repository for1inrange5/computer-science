**While Loop**

Perulangan (looping) yang berjalan **selama suatu kondisi bernilai `True`** (memenuhi syarat).

Begitu kondisinya bernilai `False` (sudah tidak memenuhi syarat), barulah perulangan tersebut berhenti.

*Contoh:* Pada `while n <= j:`, blok kode di dalamnya akan diulang terus selama nilai `n` masih kurang dari atau sama dengan `j`. Ketika nilai `n` bertambah dan melebihi `j`, kondisi menjadi `False` dan perulangan berhenti.

```python
n = 1
j = 5

while n <= j:
    print(n)
    n = n + 1
```

**Contoh Studi Kasus: Membuat PIN Acak 5 Digit** \
Kita akan membuat PIN 5 digit menggunakan perulangan `while` dan modul `random`.

---

### *Desain Algoritma*

1. **Inisialisasi**:
   - Impor modul `random` untuk menghasilkan angka acak.
   - Siapkan variabel `pin` bertipe string kosong (`""`) sebagai penampung digit PIN.
2. **Kondisi Perulangan (`while`)**:
   - Tentukan kondisi perulangan: loop berjalan **selama** panjang `pin` masih kurang dari 5 digit (`len(pin) < 5`).
3. **Proses di dalam Loop**:
   - Bangkitkan 1 angka acak antara 0 sampai 9 menggunakan `random.randint(0, 9)`.
   - Ubah angka acak tersebut menjadi format string (`str(digit)`), lalu sambungkan/gabungkan ke variabel `pin`.
4. **Terminasi & Output**:
   - Ketika panjang `pin` sudah mencapai 5 digit, kondisi `len(pin) < 5` bernilai `False` sehingga perulangan berhenti.
   - Tampilkan nilai akhir `pin` ke layar (console).

> **Catatan Logika:** Perulangan `while` bekerja dengan prinsip **"jalan selama kondisi True"**, bukan "berhenti ketika kondisi True". Oleh karena itu, kita memakai `len(pin) < 5` (selama panjangnya < 5), bukan `p <= 5` atau `len(pin) == 5`.

```python
import random

# 1. Inisialisasi variabel string kosong untuk menampung digit PIN
pin = ""

# 2. Perulangan berjalan SELAMA panjang karakter PIN masih kurang dari 5 digit
while len(pin) < 5:
    # 3. Acak 1 angka (0-9) lalu gabungkan ke dalam variabel pin
    digit = random.randint(0, 9)
    pin = pin + str(digit)

# 4. Setelah mencapai 5 digit, loop selesai dan cetak hasilnya
print(pin)
```

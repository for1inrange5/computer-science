### Modulo

modulus adalah sisa dari pembagian, kalo biasanya pembagian biasa itu $\frac{6}{3} = 2$, modulus itu sisa bagi, jadi kalo $6 \bmod 3 = 0$, kenapa? karena $\frac{6}{3}$ karena 6 bisa di bagi tanpa sisa karena $6 = 3 \cdot 2$, contoh kode di bawah

```python
print(f"hasilnya {6 // 3}, tidak ada sisa")
print(f"hasilnya {6 % 3}, karena tidak ada sisa")

# Kalo 7 bagi 3 gimana?
print(f"{7 // 3} hasilnya 2, karena pembagian biasa, karena pembagian itu tujuannya membagi sesuati menjadi bagian yang sama besar")
print(f"{7 % 3} hasilnya 1, kenapa? karena modulo itu mengambil sisa dari pembagian bulat: 7 x 2 + 1")
```

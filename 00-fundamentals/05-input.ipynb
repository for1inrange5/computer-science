#### Input
Input di program itu di butuhkan, kenapa? karena kalo program di liat doang itumah videoa wkaokwoaow \
jadi apa itu input?, input adalah proses dimana program mengambil nilai/value dari user langsung, contoh `Masukan Nilai: ` nah ini input, langsung aja liat contohnya

```python
input() # ini gak menghasilkan apa apa karena kita engga masukin ke fungsi print yang dimana itu mencetak ke konsol
print(input()) # Ini yang udah di print, tapi masih ada yang kurang, yaitu instruksi
```

```python
# kita dapat memasukan instruksi ke dalam fungsi input
print(input("Halo! selamat pagi! silahkan masukan pagi/siang: "))
```

```python
# lalu bagaimana jika kita mengkombinasikan dengan variabel?
# ini sama saja dengan mendeklarasikan variabel, namun ini di tambahin input dari user sebagai nilai
umur = input("Masukan umur anda: ")
print(umur)
```

#### Ada Catatan Penting!
input pada dasarnya dia menerima string, jadi outputnya juga berupa string

```python
# Cek tipe data input default
age = input("Your age: ")
print(type(age))
```

```python
# Namun gimana? apakah ada cara untuk mengubahnya? tentu ada!, jadi kalian hanya menambahkan int di bagian input
age = int(input("Your age: "))
print(type(age))
```

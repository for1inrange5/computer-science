#### Algoritma dan Seleksi
Algoritma, yaitu serangkaian langkah yang jelas dan berurutan buat menyelesaikan masalah.

Contohnya kayak membuat roti, kita punya masalah laper, terus menyelesaikan masalah pake cara membuat roti, nah membuat roti itu juga ada langkah langkahnya.

Seleksi, *(biasanya seleksi atau percabangan)* itu tuh cara gimana sebuah algoritma memilih jalur yang berbeda tergantung pada kondisi. kondisinya cuma *(True)* dan *(False)* dalam python kita bisa tulis `if` `elif` `else`

```python
# Contoh kode daripada algoritma dan seleksi
nama = "Panji"
nilai = 67

# Nah di bagian algortima itu mengecek kondisi
# kalo seleksi bagian if else nya, kalo gak ini ya itu, kalo gak itu ya ini

if(nilai > 80):
    print(f"Mantap {nama}")
elif(nilai < 10):
    print(f"Dongo {nama}")
elif(nilai == 67):
    print("SIKS SEVENNN!!")
else:
    print("Masukan nilai yang bener rawr")
```

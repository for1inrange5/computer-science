#### Tipe data
Santai saja ini mudah kok, cuma bedain mana yang beda aja 3 tipe data ini yang paling umum.
- Int itu bilangan bulat contoh `10, 11, 12`, gak ada koma, gak ada aneh aneh, ya bilangan bulat pada umumnya
- Float itu bilangan desimal, contoh `10.1 11.2 11.5`, ada koma, kalo udah pake float harus berhati hati ygy, karena biasanya beda dikit udah beda semua
- String itu sebuah karakter yang di deretkan kaya misal `"Panji suka Fangyi"` gitu! ga ada aneh aneh, tapi jangan lupa tanda petiknya ya!

```python
# Contoh penerapan di kode
int = 10
float = 10.1
string = "Panji suka Fangyi"

# Cek tipe data pake type
print(type(int))
print(type(float))
print(type(string))

# Kalo mau di print
print(int)
print(float)
print(string)
```

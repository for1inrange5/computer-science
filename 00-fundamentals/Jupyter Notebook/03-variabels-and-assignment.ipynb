#### Merging The Variabels
Kalo kita pake variabel, kita bisa pake ulang variabel itu kapanpun, bahkan kita bisa menggabungkanya jika tipe datanya sama

```python
# Setiap string di kasih spasi agar tidak menyatu
nama = " Panji "
universitas = " Universitas Negeri Yogyakarta "
fakultas = " Fakultas Vokasi "
prodi = " D4 Teknik Elektronika "
kelas = " D2 "

# Contoh penggabungan variabel
print(nama + universitas + fakultas + prodi + kelas)
# Contoh menggunakan kalimat
kalimat = "Pria itu bernama" + nama + "dia kuliah di" + universitas + "prodi" + prodi
print(kalimat)
```

#### Aturan Penamaan
Ini bukan aturan resmi, tapi biar mempermudah kamu dan orang orang sebagai programmer
- Nama bisa hanya huruf, angka, dan underscore
- Nama tidak dapat mulai dengan angka
- Nama juga memiliki case-sensitive
- Nama tidak bisa mirip Kata kata python seperti `print` atau `float`
Beberapa orang, format python sebagai *snake case*. menggunakan lowercase dan underscores contohnya `volume` atau `high_score`

```python
# Contoh daripada aturan penamaan
nama_depan = "Panji"
kelas = "D4"
```

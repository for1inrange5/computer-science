#### Melacak Variabel
Di sini, kita mencoba memahami variabel dan melacak variabel juga disini, jadi gaskan saja ya

```python
# Contohnya
p = 10 # variabel awal
p = p + 1 # variabel p + 1
print("Variabel p + 1 =", p)

# Contoh lagi
# Kita sudah memiliki variabel p yang bernilai 11
# Karena p sudah memiliki 11, kita bisa mengkombinasikanya lagi!
q = 10 + p # 10 + 11, kenapa? karena p menyimpan 11
# q = p Bernilai 11
print(q)

#Contoh dengan string
n = "Panji" # N memiliki nilai "Panji"
m = n + " Maxxing" # M memiliki n yang dimana "Panji" + "Maxxing"
t = m + " GG Gaming!" # t memiliki nilai m + "Gg gaming!"
print(t)
```

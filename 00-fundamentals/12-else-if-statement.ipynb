#### Else if statement
ini digunakan dengan cara berantai, alih alih kita pake if terus buat bikin kondisi, kita pake elif untuk kondisi berantai, contoh:

**If berantai**

```python
umur = 19

if umur < 10:
    print("Anda masih cukup bocil untuk menjadi rawr")

if umur > 18:
    print("Anda sudah cukup untuk mengambil keputusan")

if umur == 2000:
    print("anomali inimah")
```

**Dengan elif**

```python
if umur < 10: # Jika lebih kecil dari sepuluh maka:
    print("Yihu")
elif umur > 18: # Jika lebih dari delapan belas maka:
    print("Yihi")
else: # Jika tidak keduanya maka:
    print("Anomali")
```

**Kesimpulan** \
jadi kegunaan elif disini bukan cuma mempermudah, tapi juga biar gak kebanyakan if if an, alih alih if 3, kita gunakan elif biar bercabang

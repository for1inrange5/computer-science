#### Ekspresi boolean

ini membandingkan satu nilai ke nilai lain, pake operasi perbandingan kayak `==, !=, <, <=, >` dan `>=`, operasi perbandingan ini mengevaluasi satu nilai boolean, yaitu `True` dan `False`.

**Sorting Strings**
Komputer mengurutkan teks karakter demi karakter dengan urutan kayak gini:
1. Angka `0-9` paling duli
2. Huruf kapital `A-Z`
3. Huruf kecil `a-z`

Karena ada urutan "lebih kecil" disini artinya "muncul lebih dulu dalam urutan tersebut"

```python
boolean = "10" > "9"
print(boolean)

boolean2 = "9" > "8"
print(boolean2)

boolean3 = "1112" > "1112"
print(boolean3)
```

```python
# Contoh kecil
username = input("Pick a username: ")

# Cek kondisi usernamenya kosong apa engga
is_empty = username == ""
print("Apakah username kamu kosong? " + str(is_empty))

# Cek kondisi apakah usernamenya di pake atau tidak
is_available = username != "panji"
print("Apakah username kamu tersedia? " + str(is_available))

# Daftar nama pengguna ditampilkan berdasarkan urutan abjad
comes_first = username < "panji"
print("Apakah nama pengguna kamu ada di depan nama pengguna saya? " + str(comes_first))

has_min_chars = len(username) >= 5
print("Apakah nama kamu memiliki setidaknya 5 karakter? " + str(has_min_chars))

```

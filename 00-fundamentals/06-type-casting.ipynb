#### Type Casting
Nah ini tuh fungsinya mengkonversi tipe data lain ke tipe data lain, maksudnya gimana? gini contohnya

```python
# Contoh ada input
a = input("Masukan angka pertama: ")
b = input("Masukan angka kedua: ")

print(a + b);
```

**1 + 1 = 11? kenapa?** \
ini disebabkan input default tipe datanya string, jadi kalo string + string jadinya langsung gabung, bukan di jumlahkan, kita bisa langsung pake type casting disini

```python
a = int(input("Masukan angka a: ")) # bisa langsung pake int disini
b = int(input("Masukan angka b: ")) # lebih enak int disini karena lebih ringkas dan mudah di pahami

print(a + b) # Atau disini juga bisa
```

**Studi Kasus, Menghitung nilai rata rata** \
Disini kita di minta 3 input dan menghitung rata ratanya, menggunakan float

```python
# Menggunakan array
a = int(input("Masukan nilai pertama: "))
b = int(input("Masukan nilai kedua: "))
c = int(input("Masukan nilai ketiga: "))

nilai = [a, b, c]
print(sum(nilai) / len(nilai))
# Ini adalah solusi termudah
```

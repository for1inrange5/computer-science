#### Operator Logika

operator logika ada 3, yaitu not, and, dan or, sebenernya ada banyak tapi ini kebanyakan yang di pakai


jadi singkatnya kita bisa membandingkan 2 pernyataan, kayak gini

```python
kuliah = input("Masukan Universitas: ") # Input Kuliah
cpu = input("Masukan CPU yang digunakan mahasiswa: ") # Input CPU

if cpu == "Celeron" and kuliah == "Universitas Negeri Yogyakarta": # kalo x bener dan y bener maka true
    print("BEJIR DIA PANJI GEMINK COY") # Ini kalo true
elif cpu == "Celeron" or kuliah == "Universitas Negeri Yogyakarta": # Ini salah satu, jika salah satu true, maka true semua
    print("Maybe dia panji, tapi bukan real panji, apakah lu orang lain?")
    orang_lain = input("Ya/Tidak: ")
else: # Kalo gak keduanya
    print("Damn lu beneran orang asing")

if orang_lain == "Ya":
    print("Bejir maaf gak kenal")
```

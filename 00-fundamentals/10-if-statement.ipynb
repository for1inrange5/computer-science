#### If statement
Fungsinya buat cek kondisi, kalo misal a = b itu benar maka `True` jika sebaliknya berarti `False` udah pernah di jelasin di ekspresi boolean, jadi disini playground saja

```python
nama = "Panji"
jumlah = int(input("Masukan Angka: "))

if jumlah >= 10:
    ucapan = f"Angka sepuluh itu keren kan Bang {nama}"
    print(ucapan)

if (jumlah != 10):
    ucapan_2 = "I like computer science"
    print(ucapan_2)
```

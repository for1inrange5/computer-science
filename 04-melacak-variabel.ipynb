#### Melacak Variabel
Di sini, kita mencoba memahami variabel dan melacak variabel juga disini, jadi gaskan saja ya

```python
# Contohnya
p = 10 # variabel awal
p = p + 1 # variabel p + 1
print("Variabel p + 1 =", p)

# Contoh lagi
# Kita sudah memiliki variabel p yang bernilai 11
# Karena p sudah memiliki 11, kita bisa mengkombinasikanya lagi!
q = 10 + p # 10 + 11, kenapa? karena p menyimpan 11
# q = p Bernilai 11
print("Hasil penggabungan p + q =", q)

#Contoh dengan string
n = "Panji" # N memiliki nilai "Panji"
m = n + " Maxxing" # M memiliki n yang dimana "Panji" + "Maxxing"
t = m + " GG Gaming!" # t memiliki nilai m + "Gg gaming!"
print("Hasil Maxing =", t)
```

**Cara Kerja Variabel**

Proses assignment pada Python bekerja dengan menghubungkan nama variabel dengan sebuah objek di memori.

Ketika sebuah variabel diberikan nilai baru, nama variabel tersebut akan merujuk pada objek yang baru.

Objek sebelumnya tidak otomatis langsung dihapus. Jika sudah tidak memiliki referensi yang menggunakannya, objek tersebut dapat dibersihkan oleh mekanisme *Garbage Collection* Python sehingga memorinya dapat dibebaskan.

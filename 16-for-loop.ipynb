**For Loop**

fungsi dari pada for loop adalah mengulang, jadi yang di dalam blok itu mengulang terus sampai memenuhi range, contoh:

```python
for angka in range(5): # Untuk angka, di jangkauan 5
    print(angka) # Cetak angka
    angka = angka + 1

# hal yang menarik disini, jika kita menggunakan while
# kita wajib menginisialisasikan variabel, sedangkan angka tidak.
number = 0
while number < 5:
    print(number)
    number = number + 1
```

**For Loop**

fungsi dari pada for loop adalah mengulang, jadi yang di dalam blok itu mengulang terus sampai memenuhi range, contoh:

```python
for angka in range(5): # Untuk angka, di jangkauan 5
    print(angka) # Cetak angka

# hal yang menarik disini, jika kita menggunakan while
# kita wajib menginisialisasikan variabel, sedangkan angka tidak.
number = 0
while number < 5:
    print(number)
    number = number + 1
```

**Studi kasus: Tabungan**

Algoritma
1. Siapkan variabel tabungan = 0
2. Untuk setiap hari dari 1 sampai 30:
   2a. Tambahkan 5000 ke tabungan
   2b. Jika hari habis dibagi 10:
       Tambahkan 2000 ke tabungan
3. Tampilkan tabungan

```python
tabungan = 0 #inisialisasi tabungan

for hari in range (1, 31): # hari ke 1 ... 30, bukan 0 sampai 30
    tabungan += 5000
    if hari % 10 == 0: # hari kelipatan 10
        tabungan += 2000

print("Total tabungan budi: ", tabungan)
```
